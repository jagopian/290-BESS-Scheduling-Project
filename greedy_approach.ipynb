```python
import pandas as pd
```

```python
def parse_RT_data(filepath):
    df = pd.read_csv(filepath)
    return df['spp'][0:672].to_list()
```

```python
rt_prices = parse_RT_data('RT_PRICES_SBEAN_JAN.csv')
```

```python
def greedy(prices, eff_g, eff_ch):
    soc = 0
    previous_price = 1000
    profit = []
    pch = 5e12
    pg = 5e12
    max_soc = 5e12
    for t, price in enumerate(prices):
        if t == len(prices) - 1:
            return profit
        print(f"price: {previous_price, price, t, prices[t+1]}")
        # If the price increased from the previous timestep, commit to discharge in the next time step, else discharge
        if price > previous_price and soc >= 0 or soc >= max_soc or abs(soc - pg // 15 * (len(prices)-t-1)) < pg // 15:
            print("discharge")
            discharge = min(pg // 15, soc * eff_g // 100)
            profit.append(discharge * prices[t+1] // 1e11)
            soc -= discharge * 100 // eff_g
        else:
            print('charge')
            charge = min(pch // 15, (max_soc - soc) * 100 // eff_ch)
            profit.append(-charge * prices[t+1] // 1e11)
            soc += charge * eff_ch // 100
        previous_price = price
        print(soc/1e11)
    return profit
    
```

```python
profit = greedy(rt_prices, 93, 93)
print(sum(profit))
```
